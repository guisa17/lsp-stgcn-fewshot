released config. Only the processed LSA64 bundle is needed (about 60 MB, built locally with
`scripts/prepare_lsa64.py`); upload it to the session or point `BUNDLE` at your own storage.

```python
!git clone https://github.com/guisa17/lsp-stgcn-fewshot.git
%cd lsp-stgcn-fewshot
!pip install -q -e .
import torch
print("CUDA available:", torch.cuda.is_available())
```

```python
# Path of the processed LSA64 bundle inside this session (edit if you stored it elsewhere).
BUNDLE = "data/external/lsa64_v1.npz"
CONFIG = "configs/lsa64/hagcn_fewshot.yaml"   # or stgcn_fewshot.yaml, hagcn_fewshot_heldout_val.yaml

import os
if not os.path.isfile(BUNDLE):
    from google.colab import files
    os.makedirs(os.path.dirname(BUNDLE), exist_ok=True)
    uploaded = files.upload()  # choose lsa64_v1.npz
    os.replace(next(iter(uploaded)), BUNDLE)
```
